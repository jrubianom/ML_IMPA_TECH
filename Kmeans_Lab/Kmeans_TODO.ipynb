# Introducing Kmeans: 
Kmeans is an unsupervised model proposed to find different groups/clusters for unlabeled data.
The idea of the algorithm is simple: 
1) Choose a number of clusters $K$ and a notion of distance.
2) Choose $K$ points randomly $C = \{h_1,...,h_k\}$
3) For each point $x$ in the data, find the $h_j \in C$ closes to x. Thus, the i-th cluster will be the set of points closest to $h_i$.
4) For each cluster, redefine $h_i$ as the mean of the $i-th$ cluster.
5) Repeat 3 and 4 until convergence (if applicable)
